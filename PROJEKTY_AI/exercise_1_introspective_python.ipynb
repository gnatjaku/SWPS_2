# Exercise 1 — Introspective Python
## Python as a Self-Reflective Environment

This notebook demonstrates advanced introspection and self-modifying behavior in Python. Students explore how Python can inspect its own structures, generate functions dynamically, analyze execution, and mutate behavior during runtime.

```

import inspect
import ast
import types

def original_function(x):
    return x * 2 + 1

source = inspect.getsource(original_function)

print("=== ORIGINAL SOURCE CODE ===")
print(source)
```

```

tree = ast.parse(source)

class Transformer(ast.NodeTransformer):
    def visit_BinOp(self, node):
        return ast.BinOp(
            left=node.left,
            op=ast.Mult(),
            right=ast.Constant(value=10)
        )

transformed_tree = Transformer().visit(tree)
ast.fix_missing_locations(transformed_tree)

compiled = compile(transformed_tree, filename="<ast>", mode="exec")

namespace = {}
exec(compiled, namespace)

transformed_function = namespace['original_function']

print("=== TRANSFORMED RESULT ===")
print(transformed_function(5))
```

Discussion:
- Python becomes an environment capable of observing and modifying itself.
- Students see runtime code transformation.
- This goes beyond ordinary programming into reflective computation.
